```python
month_order = ["January", "February", "March", "April", "May", "June",
               "July", "August", "September", "October", "November", "December"]



cancel_rate= df.groupby("ArrivalDateMonth")["IsCanceled"].mean().reindex(month_order)
cancel_rate
```

```python
cancel_rate.plot(kind="bar",title="Cancellation Rate by Month")
plt.show
```

```python
sns.boxplot(data=df, x="IsCanceled",y="LeadTime")
plt.title("Lead Time vs Cancelation")
plt.show
```

```python
month_map = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12,
}

df["ArrivalDateMonthNum"] = df["ArrivalDateMonth"].map(month_map) 


sns.countplot(data=df, x="ArrivalDateMonthNum", hue="IsCanceled") #checking if there is a corelation between arrival date and cancellation
plt.title("Monthly booking Trend")
plt.show
```
